```python
import pandas as pd
```

```python
l = [1,2,3,4,5,6]
s = pd.Series(l)
s
```

```python
type(s)
```

```python
# modify indexing in rows
update_indexing = pd.Series(l , index=['i0','i1','i2','i3','i4','i5'])
update_indexing
```

```python
# Modify dtype
# using  parameter --> dtype= 
# syntax:-   dtype='datatype'
modify_dtype = pd.Series(update_indexing , dtype='str')
modify_dtype
```

```python
categorical_data = ['Car' , 'Apple' , 'Ship' , 'Banana']
s = pd.Series(categorical_data , index=['C' , 'A' , 'S' , 'B'] , name='PyPandasSeries')
s 
```

```python
numeric_data = [100 , 200 , 300 , 400]
s = pd.Series(numeric_data , index=['int0','int1','int2','int3'] , name='Pands_Series')
s
```

```python
change_index = pd.Series(numeric_data , index=['f0','f1','f2','f3'] , dtype=float)
change_index
```
