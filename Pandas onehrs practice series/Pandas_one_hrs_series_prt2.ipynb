```python
datas = {
    'Name':['JOHN' , 'ANDREW' , 'EIGEN ROSUM'],
    'Age':[29 , 20 , 23],
    'Salary':[20000.0 , 100000.0 , 450000.0],
    'Gender':1

}

df = pd.DataFrame(datas)
df
```

```python
avg_salary = df['Salary'].mean()
avg_salary
```

```python
min_salary = df['Salary'].min()
min_salary
```

```python
max_salary = df['Salary'].max()
max_salary
```

```python
sum_Salary = df['Salary'].sum()
sum_Salary
```

```python
avg_age_salary = df[['Age' , 'Salary']].mean()
avg_age_salary #Age ka avg float main diya ha q k python data ka loss nahi krta
```

```python
min_salary_age = df[['Salary' , 'Age']].min()
min_salary_age #Min age float main diya ha q k python data ka loss nahi krta
```
