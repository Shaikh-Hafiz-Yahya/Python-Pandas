Fiverr par jab aapke paas bara dataset aata hai, toh manually ek ek column check karna mushkil hota hai. Pandas ki ye 5 tricks aapka ghanton ka kaam minto mein kar dengi:

### 1. df.info() aur df.describe()
Ye do commands aapko poore data ka "X-ray" karke de deti hain.

info(): Batata hai ke kis column mein kitni missing values hain aur data types kya hain.

describe(): Numbers wale columns ka Mean, Median, Min, aur Max nikal deta hai, jis se Outliers ka foran pata chal jata hai.

### 2. df.isnull().sum()
Client ko report deni ho ke data kitna ganda hai, toh ye command use karein. Ye har column ke hisab se missing values ki tadad bata deti hai.

### 3. value_counts() (For Categorical Data)
Agar aapke paas "City" ya "Gender" ka column hai, toh df['City'].value_counts() likhne se aapko pata chal jayega ke konsi city kitni baar aayi hai. Isse spelling mistakes (e.g., "Karachi" vs "Karchi") pakri jati hain.

### 4. df.drop_duplicates(inplace=True)
Sirf ek line likhne se hazaron rows mein se double entries khatam ho jati hain. Fiverr par clients aksar "Clean list of emails" ya "Unique leads" maangte hain, wahan ye best hai.

### 5. df.apply(lambda x: ...)
Ye sab se powerful trick hai. Agar aapne poore column mein koi khas tabdeeli karni hai (e.g., har price ke saath '$' sign lagana ya text ko format karna), toh apply function loop chalaye baghair foran kaam kar deta hai.
