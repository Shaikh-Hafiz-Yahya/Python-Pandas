# Topic:  Delete and Insert DATA: 

```python
import pandas as pd
```

```python
df = pd.DataFrame({'A':[1 , 3 , 4 ,6 , 8] , 'B':[2 , 5 , 7 , 9 , 10]})
df
```

insert() method1

```python
# insert() fn
df.insert(0 , 'Python' , df['A'])  #insert column into DataFrame at specified location.
df
```

### OR  Method2

```python
df.insert(1 , 'Numpy' , [8,6,4,3,1])
df
```

```python
# CREATE NEW COLUMN AND STORE FROM COLUMN A --->  FIRTST THREE ROWS
df.insert(2 , 'Matplotlib' , df.loc[0:2 , 'A'])
df
```

OR

```python
df['ScikitLearn'] = df['A'][0:3]
df
```

### DELETE COLUMN  
USING POP()

```python
df_dic = pd.DataFrame({'A':[20 , 25 , 30 , 35] , 'B':[40 , 45 , 50 , 55] , 'C':[60 , 65, 70 , 75]})
df_dic
```

```python
df_dic.pop('C') #DELETE COLUMN C
```

```python
df_dic #after deletion
```

OR

USING del keyward

```python
del df_dic['A']
df_dic
```
